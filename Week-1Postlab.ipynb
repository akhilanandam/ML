```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")

print("First 4 Rows")
print(df.head())
```

```python
print("Last 5 Rows")
print(df.tail())

print("\nRandom 5 Rows")
print(df.sample(5))
```

```python
sns.histplot(df['age'], bins=20)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,2,1)
sns.countplot(x='sex', hue='survived', data=df)
plt.title("Gender vs Survival 24EU01107")

plt.subplot(1,2,2)
sns.countplot(x='class', hue='survived', data=df)
plt.title("Passenger Class vs Survival 24EU01107")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,4))
sns.countplot(x='survived', data=df)
plt.title("Survival Count 24EU01107")
plt.show()
plt.figure(figsize=(15,5))
sns.countplot(x='sex', hue='survived', data=df)
plt.title("Gender vs Survival 24EU01107")
plt.show()
sns.scatterplot(x='age', y='fare', hue='survived', data=df)
plt.title("Age vs Fare 24EU01107")
plt.show()
```
